```python
# Mise a l'echelle
X_train /= 255.0
X_test  /= 255.0

# Mean et std calcules sur le train seulement
mean = X_train.mean(axis=(0, 1, 2))
std  = X_train.std(axis=(0, 1, 2))

X_train = (X_train - mean) / std
X_test  = (X_test  - mean) / std
```

```python
X_train = X_train.reshape(X_train.shape[0], -1)
X_test  = X_test.reshape(X_test.shape[0], -1)

X_test.shape
```

```python
def initializer(X):
    W = np.random.randn(X.shape[1], 1)
    b = np.random.randn(1)
    return (W, b)
```

```python
def model(X, W, b):
    Z = X.dot(W) + b
    A = 1/(1 + np.exp(-Z))
    return A
```

```python
from sklearn.metrics import log_loss as sklearn_log_loss
def log_loss(A, y):
    return sklearn_log_loss(y, A)
```

```python
def gradiants(A, X, y):
    dW = 1/len(y)*np.dot(X.T, A-y)
    db = 1/len(y)*np.sum(A-y)
    return (dW, db)
```

```python
def update(dW, db, W, b, learning_rate):
    W = W - learning_rate*dW
    b = b - learning_rate*db
    return (W, b)
```

```python
def predict(X, W, b):
    A = model(X, W, b)
    return (A>=0.5, A)
```

```python
from sklearn.metrics import accuracy_score
```

```python
def artifitial_neuron(X, y, learning_rate = 0.1, n_iter = 100):
    #Initialization
    W, b = initializer(X)

    Loss = []

    for i in range(n_iter):
        A = model(X, W, b)
        Loss.append(log_loss(A, y))
        dW, db = gradiants(A, X, y)
        W, b = update(dW, db, W, b, learning_rate)

    y_pred, y_pred_prob = predict(X, W, b)

    print(accuracy_score(y, y_pred))

    plt.plot(Loss)
    plt.show()

    return (W, b)
```

```python
W, b = artifitial_neuron(X_train, y_train, n_iter=2000)
```

```python
y_pred, A = predict(X_test, W, b)

loss = log_loss(A, y_test)
loss
```
